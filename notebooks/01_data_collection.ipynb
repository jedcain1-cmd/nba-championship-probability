```python
import time

def get_team_season_data(end_year):
    """
    Collect Basketball Reference advanced team data for one NBA season.

    Example:
    end_year=1986 retrieves the 1985-86 season.
    """

    url = f"https://www.basketball-reference.com/leagues/NBA_{end_year}.html"
    tables = pd.read_html(url)

    # Find the advanced team statistics table automatically
    df = None

    for table in tables:
        if table.columns.nlevels == 2:
            bottom_level = [str(col[1]) for col in table.columns]

            if (
                "Team" in bottom_level
                and "ORtg" in bottom_level
                and "DRtg" in bottom_level
                and "SRS" in bottom_level
            ):
                df = table.copy()
                break

    if df is None:
        raise ValueError(
            f"Advanced team statistics table not found for {end_year}"
        )

    # Preserve offense/defense labels when flattening headers
    new_columns = []

    for top, bottom in df.columns:
        if top == "Offense Four Factors":
            new_columns.append("OFF_" + bottom)
        elif top == "Defense Four Factors":
            new_columns.append("DEF_" + bottom)
        elif bottom and not str(bottom).startswith("Unnamed"):
            new_columns.append(bottom)
        else:
            new_columns.append(top)

    df.columns = new_columns

    # Remove separator columns
    df = df.loc[:, ~df.columns.str.startswith("Unnamed")]

    # Remove playoff asterisk from team names
    df["Team"] = (
        df["Team"]
        .str.replace("*", "", regex=False)
        .str.strip()
    )

    # Remove League Average row
    df = df[df["Team"] != "League Average"].copy()

    # Add season identifiers
    df["Season_End"] = end_year
    df["Season"] = f"{end_year - 1}-{str(end_year)[-2:]}"

    return df
```

```python
all_seasons = []
failed_seasons = []

# 1986 = 1985-86 season
# 2026 = 2025-26 season
for end_year in range(1986, 2027):

    season_name = f"{end_year - 1}-{str(end_year)[-2:]}"
    print(f"Collecting {season_name}...")

    try:
        season_df = get_team_season_data(end_year)

        all_seasons.append(season_df)

        # Save an individual backup for this season
        season_df.to_csv(
            f"../data/team_stats_{season_name}.csv",
            index=False
        )

        print(f"✓ {season_name}: {len(season_df)} teams")

    except Exception as e:
        failed_seasons.append((season_name, str(e)))
        print(f"✗ {season_name} FAILED: {e}")

    # Be respectful of Basketball Reference's request limits
    time.sleep(4)

print("\nCollection finished.")
print("Successful seasons:", len(all_seasons))
print("Failed seasons:", len(failed_seasons))
```

```python
def get_playoff_games(end_year):
    """
    Collect every NBA playoff game for one season from Basketball Reference.

    Example:
    end_year=1986 retrieves the 1985-86 playoffs.
    """

    url = f"https://www.basketball-reference.com/playoffs/NBA_{end_year}_games.html"

    tables = pd.read_html(url)
    df = tables[0].copy()

    # Keep only the fields needed for modeling
    df = df[
        ["Date", "Visitor/Neutral", "PTS", "Home/Neutral", "PTS.1"]
    ].copy()

    # Rename columns
    df.columns = [
        "Date",
        "Away_Team",
        "Away_PTS",
        "Home_Team",
        "Home_PTS"
    ]

    # Convert scores to numeric
    df["Away_PTS"] = pd.to_numeric(df["Away_PTS"], errors="coerce")
    df["Home_PTS"] = pd.to_numeric(df["Home_PTS"], errors="coerce")

    # Remove any rows that are not completed games
    df = df.dropna(subset=["Away_PTS", "Home_PTS"]).copy()

    # Convert scores to integers
    df["Away_PTS"] = df["Away_PTS"].astype(int)
    df["Home_PTS"] = df["Home_PTS"].astype(int)

    # Add season identifiers
    df["Season_End"] = end_year
    df["Season"] = f"{end_year - 1}-{str(end_year)[-2:]}"

    # Create modeling target
    # 1 = home team won
    # 0 = away team won
    df["Home_Win"] = (
        df["Home_PTS"] > df["Away_PTS"]
    ).astype(int)

    # Arrange columns
    df = df[
        [
            "Season",
            "Season_End",
            "Date",
            "Away_Team",
            "Away_PTS",
            "Home_Team",
            "Home_PTS",
            "Home_Win"
        ]
    ]

    return df
```

```python
all_playoff_seasons = []
failed_playoff_seasons = []

# 1985-86 through 2024-25
for end_year in range(1986, 2026):

    season_name = f"{end_year - 1}-{str(end_year)[-2:]}"
    print(f"Collecting {season_name} playoffs...")

    try:
        playoff_df = get_playoff_games(end_year)

        all_playoff_seasons.append(playoff_df)

        # Save individual season backup
        playoff_df.to_csv(
            f"../data/playoff_games_{season_name}.csv",
            index=False
        )

        print(f"✓ {season_name}: {len(playoff_df)} games")

    except Exception as e:
        failed_playoff_seasons.append((season_name, str(e)))
        print(f"✗ {season_name} FAILED: {e}")

    # Respect Basketball Reference request limits
    time.sleep(4)

print("\nCollection finished.")
print("Successful playoff seasons:", len(all_playoff_seasons))
print("Failed playoff seasons:", len(failed_playoff_seasons))
```

```python
# Combine all playoff seasons
all_playoff_games = pd.concat(
    all_playoff_seasons,
    ignore_index=True
)

# Convert Date to datetime
all_playoff_games["Date"] = pd.to_datetime(
    all_playoff_games["Date"]
)

# Sort chronologically
all_playoff_games = all_playoff_games.sort_values(
    ["Season_End", "Date"]
).reset_index(drop=True)

print("=== BASIC CHECKS ===")
print("Total playoff games:", len(all_playoff_games))
print("Total seasons:", all_playoff_games["Season"].nunique())
print("First season:", all_playoff_games["Season"].iloc[0])
print("Last season:", all_playoff_games["Season"].iloc[-1])

# Check duplicate games
duplicates = all_playoff_games.duplicated(
    subset=[
        "Season",
        "Date",
        "Away_Team",
        "Home_Team"
    ]
).sum()

print("Duplicate games:", duplicates)

# Check missing values
print("\n=== MISSING VALUES ===")
print(all_playoff_games.isna().sum())

# Home-court win rate
print("\n=== HOME WIN RATE ===")
print(all_playoff_games["Home_Win"].value_counts())
print(
    "Home win percentage:",
    round(all_playoff_games["Home_Win"].mean(), 3)
)

# Games per postseason
print("\n=== GAMES PER POSTSEASON ===")
games_by_season = (
    all_playoff_games
    .groupby("Season")
    .size()
)

print(games_by_season.to_string())

# Final game from first and last completed postseason
print("\n=== FIRST POSTSEASON FINAL GAME ===")
display(
    all_playoff_games[
        all_playoff_games["Season"] == "1985-86"
    ].tail(1)
)

print("\n=== MOST RECENT POSTSEASON FINAL GAME ===")
display(
    all_playoff_games[
        all_playoff_games["Season"] == "2024-25"
    ].tail(1)
)
```

```python
# Select the team-season variables we initially want
team_features = all_team_seasons[
    [
        "Season",
        "Team",
        "W_PCT",
        "SRS",
        "ORtg",
        "DRtg",
        "NRtg",
        "Pace"
    ]
].copy()

# Create home-team version
home_features = team_features.rename(
    columns={
        "Team": "Home_Team",
        "W_PCT": "Home_W_PCT",
        "SRS": "Home_SRS",
        "ORtg": "Home_ORtg",
        "DRtg": "Home_DRtg",
        "NRtg": "Home_NRtg",
        "Pace": "Home_Pace"
    }
)

# Merge home-team regular-season stats onto each playoff game
model_data = all_playoff_games.merge(
    home_features,
    on=["Season", "Home_Team"],
    how="left"
)

print("Playoff games:", len(model_data))
print("\nMissing home-team matches:")
print(model_data["Home_SRS"].isna().sum())
```

```python
# Create away-team version
away_features = team_features.rename(
    columns={
        "Team": "Away_Team",
        "W_PCT": "Away_W_PCT",
        "SRS": "Away_SRS",
        "ORtg": "Away_ORtg",
        "DRtg": "Away_DRtg",
        "NRtg": "Away_NRtg",
        "Pace": "Away_Pace"
    }
)

# Merge away-team regular-season stats onto each playoff game
model_data = model_data.merge(
    away_features,
    on=["Season", "Away_Team"],
    how="left"
)

print("Playoff games:", len(model_data))

print("\nMissing home-team matches:")
print(model_data["Home_SRS"].isna().sum())

print("\nMissing away-team matches:")
print(model_data["Away_SRS"].isna().sum())
```

```python
# Create home-minus-away feature differences

model_data["W_PCT_Diff"] = (
    model_data["Home_W_PCT"] - model_data["Away_W_PCT"]
)

model_data["SRS_Diff"] = (
    model_data["Home_SRS"] - model_data["Away_SRS"]
)

model_data["ORtg_Diff"] = (
    model_data["Home_ORtg"] - model_data["Away_ORtg"]
)

model_data["DRtg_Diff"] = (
    model_data["Home_DRtg"] - model_data["Away_DRtg"]
)

model_data["NRtg_Diff"] = (
    model_data["Home_NRtg"] - model_data["Away_NRtg"]
)

model_data["Pace_Diff"] = (
    model_data["Home_Pace"] - model_data["Away_Pace"]
)

print(
    model_data[
        [
            "W_PCT_Diff",
            "SRS_Diff",
            "ORtg_Diff",
            "DRtg_Diff",
            "NRtg_Diff",
            "Pace_Diff",
            "Home_Win"
        ]
    ].isna().sum()
)
```
